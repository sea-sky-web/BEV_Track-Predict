## 模型优化与轻量化

为了实现模型轻量化，可以尝试以下方法：
1. 使用较小的backbone网络（如MobileNetV3、ResNet18等）
2. 减少特征通道数
3. 使用模型压缩技术（如剪枝、量化）
4. 优化网络架构设计

以下是一些可能的优化示例：

```python
# 优化后的模型创建函数
def create_lightweight_model(num_views, proj_mats, reduced_hw, feat_hw, device):
    """创建轻量化MVDet模型"""
    from models import MVDetLikeNet
    from torchvision.models import mobilenet_v3_large
    
    # 可以在这里替换backbone为MobileNetV3
    model = MVDetLikeNet(
        num_views=num_views,
        proj_mats=proj_mats,
        reduced_hw=reduced_hw,
        feat_hw=feat_hw,
        feat_ch=256,  # 减少特征通道数
        pretrained=True,
        add_coord=True,
    ).to(device)
    
    print(f"轻量化模型已创建，特征通道数: 256")
    return model

# 打印模型参数量
def count_parameters(model):
    total_params = sum(p.numel() for p in model.parameters())
    trainable_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
    print(f"总参数量: {total_params / 1e6:.2f}M")
    print(f"可训练参数量: {trainable_params / 1e6:.2f}M")
    return total_params, trainable_params
```
